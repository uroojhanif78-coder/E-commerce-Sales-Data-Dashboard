# Revenue Distribution


````python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Revenue",
    bins=50,
    kde=True
)

plt.title("Revenue Distribution")
plt.show()
````

To make the graph more readable:

````python
price_limit = df["Revenue"].quantile(0.99)

sns.histplot(
    data=df[df["Revenue"] <= price_limit],
    x="Revenue",
    bins=50,
    kde=True
)

plt.title("Revenue Distribution — Up to 99th Percentile")
plt.show()
````

# BIVARIATE ANALYSIS

# 1.Revenue by category


````python
category_sales = (
    df.groupby("Category")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

category_sales
````

# Visualization

````python
plt.figure(figsize=(11, 6))

sns.barplot(
    x=category_sales.values,
    y=category_sales.index
)

plt.title("Revenue by Category")
plt.xlabel("Total Revenue")
plt.ylabel("Category")
plt.show()
````

# 2.Average Order Value by Region


````python
region_value = (
    df.groupby("Region")["Revenue"]
      .mean()
      .sort_values(ascending=False)
)

region_value
````

````python
sns.barplot(
    x=region_value.values,
    y=region_value.index
)

plt.title("Average Order Value by Region")
plt.xlabel("Average Revenue")
plt.ylabel("Region")
plt.show()
````

# 3.Units_Sold vs Revenue

````python
sample_df = df.sample(
    min(5000, len(df)),
    random_state=42
)

sns.scatterplot(
    data=sample_df,
    x="Revenue",
    y="Units_Sold",
    alpha=0.5
)

plt.title("Revenue vs Units Sold")
plt.show()
````

# MULTIVARIATE ANALYSIS

````python
top_cats = (
    df["Category"]
    .value_counts()
    .head(5)
    .index
)

subset = df[df["Category"].isin(top_cats)]

sns.boxplot(
    data=subset,
    x="Revenue",
    y="Category",
    hue="Region"
)

plt.title("Revenue by Category and Region")
plt.show()
````
